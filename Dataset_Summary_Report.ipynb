# Dataset Summary Report

## Objective
This notebook uses **Python and Pandas** to load and explore a student CSV dataset. It demonstrates how to inspect the dataset's structure, data types, sample records, descriptive statistics, and missing values.

### Analysis covered
- Loading a CSV file into a Pandas DataFrame
- Inspecting the first and last rows
- Viewing random samples
- Checking shape, columns, and index
- Checking data types and dataset information
- Generating descriptive statistics
- Identifying missing values
- Recording basic observations about the dataset


```python
import pandas as pd
import numpy as np

# Load the CSV dataset
file_path = "Day7_Student_Dataset.csv"
df = pd.read_csv(file_path)

print("Dataset loaded successfully.")
print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")
```

## 1. First Few Rows

```python
# Display the first 5 rows
df.head()
```

## 2. Last Few Rows

```python
# Display the last 5 rows
df.tail()
```

## 3. Random Sample

```python
# Display 5 randomly selected rows
df.sample(5, random_state=42)
```

## 4. Dataset Shape

```python
# Shape returns (number of rows, number of columns)
print("Shape:", df.shape)
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

## 5. Columns

```python
# List all column names
print("Columns:")
for column in df.columns:
    print("-", column)
```

## 6. Index

```python
# Inspect the DataFrame index
print(df.index)
```

## 7. Data Types

```python
# Check the data type of every column
df.dtypes
```

## 8. Dataset Information

```python
# info() shows column names, non-null counts, and data types
df.info()
```

## 9. Descriptive Statistics

```python
# Descriptive statistics for numerical columns
df.describe()
```

### Descriptive statistics including categorical columns

`include="all"` provides a broader summary. For categorical columns, Pandas reports values such as the number of unique values, the most frequent value, and its frequency.


```python
df.describe(include="all").T
```

## 10. Missing Values

```python
# Count missing values in each column
missing_values = df.isnull().sum()
missing_values
```

```python
# Show only columns that contain missing values
missing_values[missing_values > 0]
```

## 11. Basic Dataset Observations

```python
# Generate a compact summary that can be used for observations

print(f"• The dataset contains {df.shape[0]} rows and {df.shape[1]} columns.")
print("• Available features:", ", ".join(df.columns))
print("\n• Data types:")
print(df.dtypes.to_string())

print("\n• Missing values:")
missing = df.isnull().sum()
if missing.sum() == 0:
    print("No missing values were found.")
else:
    print(missing[missing > 0].to_string())
```

## 12. Summary and Observations

Based on the exploration:

- The dataset contains **25 rows and 7 columns**.
- The available features are **Student_ID, Name, Age, Course, Marks, Attendance_Percentage, and City**.
- `Student_ID` and `Age` are integer (`int64`) columns.
- `Marks` and `Attendance_Percentage` are floating-point (`float64`) columns.
- `Name`, `Course`, and `City` are categorical/text columns represented as `object`.
- Missing values are present in **Marks, Attendance_Percentage, and City**, with **one missing value in each**.
- The dataset can therefore be used for basic student-data exploration, while missing values should be considered before performing further analysis or modeling.
